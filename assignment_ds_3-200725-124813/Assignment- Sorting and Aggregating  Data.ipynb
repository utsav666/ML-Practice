####  `Create the following features using the Datetime feature`

- day of month
- month
- day of year
- hour

---

```python
# WRITE YOUR CODE HERE
data["day of month"]=data.Datetime.apply(lambda x: x.day_name())
data["moth"]=data.Datetime.apply(lambda x: x.month_name())
data["day of year"]=data.Datetime.apply(lambda x: x.dayofyear)
data["hour"]=data.Datetime.apply(lambda x: x.hour)
```

#### `READ THE BIG MART SALES DATA`

---

```python
big_mart_data = pd.read_csv('dataset/big_mart_sales.csv')
```

```python
big_mart_data.head()
```

#### Find out the number of unique `Item_Type` in each of the `Outlet_Identifier`

---

```python
# WRITE YOUR CODE HERE
result = big_mart_data.groupby('Outlet_Identifier')['Item_Type'].nunique().reset_index()
result.columns = ['Outlet_Identifier', 'Unique_Item_Type_Count']
```

```python
result
```

#### Find out the average `Item_Outlet_Sales` for each category of `Outlet_Size`

---

```python
big_mart_data
```

```python
# WRITE YOUR CODE HERE
result = big_mart_data.groupby('Outlet_Size')['Item_Outlet_Sales'].mean().reset_index()
result.columns = ['Outlet_Size', 'Average_Item_Outlet_Sales']
```

```python
result
```

#### Select only the categorical variables from the dataframe and use `get_dummies` function to convert those into the numerical ones. 

---

```python
# WRITE YOUR CODE HERE
result = pd.get_dummies(big_mart_data,columns=["Outlet_Size"])
```
