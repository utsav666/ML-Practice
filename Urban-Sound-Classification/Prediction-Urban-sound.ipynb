```python
mfc=[]
chr=[]
me=[]
ton=[]
lab=[]
s_rate=16000
for i in tqdm(range(len(test_data))):
    f_name='C:\\Users\\utsav\\Downloads\\UrbanSound8K.tar\\test\\Test\\'+str(test_data.ID[i])+'.wav'
    X, s_rate = librosa.load(f_name, res_type='kaiser_fast')
    mf = np.mean(librosa.feature.mfcc(y=X, sr=s_rate).T,axis=0)
    mfc.append(mf)
    #l=test_data.Class[i]
    #lab.append(l)
    try:
        t =    np.mean(librosa.feature.tonnetz(
                       y=librosa.effects.harmonic(X),
                       sr=s_rate).T,axis=0)
        ton.append(t)
    except:
        print(f_name)  
    m = np.mean(librosa.feature.melspectrogram(X, sr=s_rate).T,axis=0)
    me.append(m)
    s = np.abs(librosa.stft(X))
    c = np.mean(librosa.feature.chroma_stft(S=s, sr=s_rate).T,axis=0)
    chr.append(c)
```

```python
mfcc = pd.DataFrame(mfc)
mfcc.to_csv('mfc_test.csv', index=False)
chrr = pd.DataFrame(chr)
chrr.to_csv('chr_test.csv', index=False)
mee = pd.DataFrame(me)
mee.to_csv('me_test.csv', index=False)
tonn = pd.DataFrame(ton)
```
