# Loading dataset

```python
from datasets import load_dataset

ds = load_dataset("wikimedia/wikipedia", "20231101.hi")
```

# selecting a sample of 1k

```python
ds=ds["train"].select(range(1000))
```

# train test splitting

```python
dataset = ds.train_test_split(0.1)
print(dataset)
```

# Use an existing tokenizer to train

```python
batch_size = 100
all_texts_train = [dataset["train"][i : i + batch_size]["text"] for i in range(0, len(dataset["train"]), batch_size)]
```

```python
len(all_texts_train)
```

```python
def batch_iterator():
    for i in range(0, len(dataset), batch_size):
        yield dataset["train"][i : i + batch_size]["text"]
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("gpt2")
```

```python
tokenizer.is_fast
```

```python
new_tokenizer = tokenizer.train_new_from_iterator(batch_iterator(), vocab_size=25000)
```

# result of new tokenizer

```python
new_tokenizer(dataset["test"][:5]["text"])
```
