# Look at the top 50 mentions in the dataset (mentions in this dataset refers to the @s)

```python
import re
import pandas as pd
from collections import Counter
mention_list = []
def top_50_mention(text):
    pattern = r'(?<!\w)@s\w+'
    words_starting_with_at = re.findall(pattern, text)
    mention_list.extend(words_starting_with_at)

data["text"].apply(top_50_mention)
def top_50_most_occurred_mention(word_list):
    word_counter = Counter(word_list)
    top_50_words = word_counter.most_common(50)
    return top_50_words

# Find top 50 most occurred mention
top_50_occ_men = top_50_most_occurred_mention(mention_list)

# Create DataFrame
df = pd.DataFrame(top_50_occ_men, columns=['Word', 'Count'])
print(df)
```

# Find the top 50 most frequently used hashtags (#)

```python
import re
import pandas as pd
from collections import Counter
mention_list = []
def top_50_mention(text):
    pattern = r'(?<!\w)#\w+'
    words_starting_with_at = re.findall(pattern, text)
    mention_list.extend(words_starting_with_at)

data["text"].apply(top_50_mention)
def top_50_most_occurred_hashtag(word_list):
    word_counter = Counter(word_list)
    top_50_words = word_counter.most_common(50)
    return top_50_words

# Find top 50 most occurred mention
top_50_occ_hastag = top_50_most_occurred_hashtag(mention_list)

# Create DataFrame
df = pd.DataFrame(top_50_occ_hastag, columns=['hashtag', 'Count'])
print(df)
```

# Rule 1 - Extracting the sentences with President's mention

```python
# importing spacy and its matcher function
import spacy
from spacy.matcher import Matcher

# load english language model
nlp = spacy.load('en_core_web_sm',disable=['ner','textcat'])
```

```python
# function to find sentences mentioning president
def find_names(text):

    names = []

    # spacy doc
    doc = nlp(text)

    # pattern
    pattern = [{'LOWER':'prime'},
              {'LOWER':'minister'},
              {'POS':'ADP','OP':'?'},
              {'POS':'PROPN'}]

    # Matcher class object
    matcher = Matcher(nlp.vocab)
    matcher.add("names",[pattern])

    matches = matcher(doc)

    # finding patterns in the text
    for i in range(0,len(matches)):

        # match: id, start, end
        token = doc[matches[i][1]:matches[i][2]]
        # append token to list
        names.append(str(token))

    return names
```

```python
# extracting sentences with prime minister mentions
extracted_sentences = []
for i in range(data.shape[0]):
    extracted_sentences.append(find_names(data['text'][i]))

# # storing the extracted sentences in the dataframe
data['Prime minister'] = extracted_sentences
```

# Pattern using preposition

```python
# rule 5 function
def rule5(text):

    doc = nlp(text)

    sent = []

    for token in doc:

        # look for prepositions
        if token.pos_=='ADP':

            phrase = ''

            # if its head word is a noun
            if token.head.pos_=='NOUN':

                # append noun and preposition to phrase
                phrase += token.head.text
                phrase += ' '+token.text

                # check the nodes to the right of the preposition
                for right_tok in token.rights:
                    # append if it is a noun or proper noun
                    if (right_tok.pos_ in ['NOUN','PROPN']):
                        phrase += ' '+right_tok.text

                if len(phrase)>2:
                    sent.append(phrase)

    return sent
```

```python
data = data.drop(columns=['President_Names'])
```

```python
# create a df containing sentence and its output for rule 5
row_list = []

for i in range(len(data)):

    sent = data.loc[i,'text']
    idx = data.loc[i,'Unnamed: 0']

    # rule
    output = rule5(sent)

    dict1 = {'index':idx,'text':sent,'Output':output}
    row_list.append(dict1)

df_rule = pd.DataFrame(row_list)
```

```python
# select non-empty outputs
df_show = pd.DataFrame(columns=df_rule.columns)

for row in range(len(df_rule)):

    if len(df_rule.loc[row,'Output'])!=0:
        df_show = df_show.append(df_rule.loc[row,:])

# reset the index
df_show.reset_index(inplace = True, drop = True)
```

```python
# first 5 rows
df_show.head()
```

```python
# separate noun, preposition and noun

prep_dict = dict()
dis_dict = dict()
dis_list = []

# iterating over all the sentences
for i in range(len(df_show)):

    # sentence containing the output
    sentence = df_show.loc[i,'text']
    # year of the sentence
    year = df_show.loc[i,'index']
    # output of the sentence
    output = df_show.loc[i,'Output']

    # iterating over all the outputs from the sentence
    for sent in output:

        # separate subject, verb and object
        n1, p, n2 = sent.split()[0], sent.split()[1], sent.split()[2:]

        # append to list, along with the sentence
        dis_dict = {'text':sentence,'Year':year,'Noun1':n1,'Preposition':p,'Noun2':n2}
        dis_list.append(dis_dict)

        # counting the number of sentences containing the verb
        prep = sent.split()[1]
        if prep in prep_dict:
            prep_dict[prep]+=1
        else:
            prep_dict[prep]=1

df_sep= pd.DataFrame(dis_list)
```
